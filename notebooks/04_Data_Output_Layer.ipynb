```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("SUCCESS: All imports successful!")
```

## 1. Load Processed Data

```python
# Load processed data
df = pd.read_csv("../data/processed_data.csv")

print(f"Loaded {len(df)} records for visualization")
df.head()
```

## 2. Story: Mobile Device Market Analysis

### Our Story:
We analyze the mobile device market to understand:
- Price distribution across brands and categories
- Feature relationships (RAM, Storage, Screen Size)
- Market segmentation (Budget, Mid-range, Premium)
- Brand positioning and value propositions

## 3. Visualizations

### 3.1 Price Distribution by Brand

```python
if 'price' in df.columns and 'brand' in df.columns:
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df, x='brand', y='price')
    plt.title('Price Distribution by Brand', fontsize=16, fontweight='bold')
    plt.xlabel('Brand', fontsize=12)
    plt.ylabel('Price ($)', fontsize=12)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    # Summary statistics
    print("\nAverage Price by Brand:")
    print(df.groupby('brand')['price'].agg(['mean', 'median', 'count']).round(2))
```

### 3.2 Price vs Features Scatter Plot

```python
if 'price' in df.columns and 'ram' in df.columns:
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    
    # Price vs RAM
    axes[0].scatter(df['ram'], df['price'], alpha=0.6, s=100)
    axes[0].set_xlabel('RAM (GB)', fontsize=12)
    axes[0].set_ylabel('Price ($)', fontsize=12)
    axes[0].set_title('Price vs RAM', fontsize=14, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    
    # Price vs Storage
    if 'storage' in df.columns:
        axes[1].scatter(df['storage'], df['price'], alpha=0.6, s=100, color='orange')
        axes[1].set_xlabel('Storage (GB)', fontsize=12)
        axes[1].set_ylabel('Price ($)', fontsize=12)
        axes[1].set_title('Price vs Storage', fontsize=14, fontweight='bold')
        axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

### 3.3 Category Distribution

```python
if 'category' in df.columns:
    # Pie chart
    category_counts = df['category'].value_counts()
    
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    # Pie chart
    axes[0].pie(category_counts.values, labels=category_counts.index, autopct='%1.1f%%', startangle=90)
    axes[0].set_title('Device Category Distribution', fontsize=14, fontweight='bold')
    
    # Bar chart
    axes[1].bar(category_counts.index, category_counts.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
    axes[1].set_xlabel('Category', fontsize=12)
    axes[1].set_ylabel('Count', fontsize=12)
    axes[1].set_title('Device Count by Category', fontsize=14, fontweight='bold')
    axes[1].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
    
    print("\nCategory Statistics:")
    print(category_counts)
```

### 3.4 Interactive Visualization with Plotly

```python
if 'price' in df.columns and 'brand' in df.columns and 'screen_size' in df.columns:
    # Interactive scatter plot
    fig = px.scatter(df, 
                     x='screen_size', 
                     y='price', 
                     color='brand',
                     size='ram' if 'ram' in df.columns else None,
                     hover_data=['device_name', 'os'],
                     title='Interactive: Price vs Screen Size by Brand',
                     labels={'screen_size': 'Screen Size (inches)', 'price': 'Price ($)'})
    
    fig.update_layout(height=600, width=1000)
    fig.show()
```

### 3.5 Correlation Heatmap

```python
# Select numeric columns for correlation
numeric_cols = df.select_dtypes(include=[np.number]).columns
if len(numeric_cols) > 1:
    corr_matrix = df[numeric_cols].corr()
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
                square=True, linewidths=1, cbar_kws={"shrink": 0.8})
    plt.title('Feature Correlation Heatmap', fontsize=16, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()
```

## 4. Key Findings and Insights

### Summary of Analysis:

```python
print("KEY FINDINGS:\n")
print("=" * 50)

if 'price' in df.columns:
    print(f"\nPrice Analysis:")
    print(f"  - Average Price: ${df['price'].mean():.2f}")
    print(f"  - Price Range: ${df['price'].min():.2f} - ${df['price'].max():.2f}")
    print(f"  - Median Price: ${df['price'].median():.2f}")

if 'brand' in df.columns and 'price' in df.columns:
    print(f"\nBrand Analysis:")
    brand_avg = df.groupby('brand')['price'].mean().sort_values(ascending=False)
    print(f"  - Most Expensive Brand: {brand_avg.index[0]} (${brand_avg.iloc[0]:.2f})")
    print(f"  - Most Affordable Brand: {brand_avg.index[-1]} (${brand_avg.iloc[-1]:.2f})")

if 'category' in df.columns:
    print(f"\nMarket Segmentation:")
    print(f"  - Categories: {', '.join(df['category'].unique())}")
    print(f"  - Distribution: {df['category'].value_counts().to_dict()}")

print("\n" + "=" * 50)
```
