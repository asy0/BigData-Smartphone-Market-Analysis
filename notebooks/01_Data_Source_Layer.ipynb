## 1. Setup and Imports

```python
import sys
import os
sys.path.append('..')

import pandas as pd
import numpy as np
import requests
from datetime import datetime
import json

# Project modules
from src.data_sources import (
    collect_mobile_api_data,
    load_kaggle_dataset,
    merge_data_sources,
    validate_data_quality
)
from src.mobile_api_example import search_device

print("SUCCESS: All imports successful!")
```

## 2. Data Source 1: Mobile API

We use the Mobile API to fetch device specifications and information.

```python
# List of devices to query
device_names = [
    "iPhone 15 Pro",
    "Samsung Galaxy S24",
    "Google Pixel 8",
    "OnePlus 12",
    "Xiaomi 14"
]

# Collect data from API
api_data = collect_mobile_api_data(device_names)

print(f"Collected {len(api_data)} devices from Mobile API")
print("\nSample data:")
api_data.head()
```

### API Data Structure Documentation

```python
if len(api_data) > 0:
    print("Data Structure:")
    print(f"Columns: {list(api_data.columns)}")
    print(f"\nData Types:")
    print(api_data.dtypes)
    print(f"\nSample Record:")
    if 'api_data' in api_data.columns:
        print(json.dumps(api_data.iloc[0]['api_data'], indent=2))
```

## 3. Data Source 2: Kaggle Dataset

We load a mobile device dataset from Kaggle.

```python
# Load Kaggle dataset
kaggle_df = load_kaggle_dataset("../data/mobile_devices.csv")

print(f"Loaded {len(kaggle_df)} records from Kaggle dataset")
print(f"\nColumns: {list(kaggle_df.columns)}")
kaggle_df.head()
```

### Kaggle Dataset Metadata

```python
print("Dataset Info:")
print(kaggle_df.info())
print("\n\nDataset Statistics:")
kaggle_df.describe()
```

## 4. Merge Data Sources

Combine data from API and Kaggle into a unified dataset.

```python
# Merge all data sources (API and Kaggle)
merged_df = merge_data_sources(api_data, kaggle_df)

print(f"SUCCESS: Total merged records: {len(merged_df)}")
print(f"Columns: {list(merged_df.columns)}")
merged_df.head(10)
```

## 5. Merge Data Sources

Combine data from all sources into a unified dataset.

```python
# Validate data quality
quality_report = validate_data_quality(merged_df)

print("Data Quality Report:")
print(f"Total Records: {quality_report['total_records']}")
print(f"Duplicate Records: {quality_report['duplicate_records']}")
print("\nMissing Values:")
for col, count in quality_report['missing_values'].items():
    if count > 0:
        pct = quality_report['missing_percentage'][col]
        print(f"  {col}: {count} ({pct:.2f}%)")
```

## 6. Data Quality Assessment

Check data quality: completeness, validity, and inconsistencies.

```python
# Validate data quality
quality_report = validate_data_quality(merged_df)

print("Data Quality Report:")
print(f"Total Records: {quality_report['total_records']}")
print(f"Duplicate Records: {quality_report['duplicate_records']}")
print("\nMissing Values:")
for col, count in quality_report['missing_values'].items():
    if count > 0:
        pct = quality_report['missing_percentage'][col]
        print(f"  {col}: {count} ({pct:.2f}%)")
```

## 7. Save Raw Data

Save the collected data for further processing.

```python
# Statistical summary
if 'price' in merged_df.columns:
    print("Price Statistics:")
    print(merged_df['price'].describe())
    
if 'brand' in merged_df.columns:
    print("\n\nBrand Distribution:")
    print(merged_df['brand'].value_counts())
    
if 'os' in merged_df.columns:
    print("\n\nOperating System Distribution:")
    print(merged_df['os'].value_counts())
```

## 8. Save Raw Data

Save the collected data for further processing.

```python
# Save to CSV for next layer
output_path = "../data/raw_merged_data.csv"
merged_df.to_csv(output_path, index=False)
print(f"SUCCESS: Raw data saved to {output_path}")
```
